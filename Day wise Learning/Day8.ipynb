## Decision Tree

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data={
    'Hours_studied':[1,2,3,4,5,6,7,8],
    'Attedance':[40,50,60,70,80,90,85,95],
    "Pass":[0,0,0,1,1,1,1,1]
}
df=pd.DataFrame(data)
X=df[['Hours_studied','Attedance']]
y=df[['Pass']]
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
model=DecisionTreeClassifier(criterion='gini', max_depth=3)
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
print("Predicted Result:",y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:",accuracy)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plot_tree(
    model,
    feature_names=['hours_studies','attendence'],
    class_names=['Fail','Pass'],
    filled=True
)
plt.show()
```

## **Random Forest**

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
data=load_iris()
X = data.data
y = data.target
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    bootstrap=True,
    random_state=42
)
rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
print("Prediction:",y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:",accuracy)
```

```python
plt.figure(figsize=(12,6))
plot_tree(
    rf.estimators_[0], # First decision tree
    feature_names=data.feature_names,
    class_names=data.target_names,
    filled=True,
    rounded=True
)
plt.title("One Decision Tree from Random Forest (Iris Dataset)")
plt.show()
```

```python
# Get Feature importance

importances = rf.feature_importances_
features = data.feature_names
```

```python
plt.figure()
plt.bar(features, importances)
plt.ylabel("Importances Score")
plt.xlabel("Features imp_random forest")
plt.xticks(rotation=45)
plt.show()
```
