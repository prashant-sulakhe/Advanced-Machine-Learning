```python
res=int(input("Enter your result:"))
if(res>=35 and res<50):
  print("Third class")
elif(res>=50 and res<60):
  print("Second class")
elif(res>=60 and res<80):
  print("First class")
elif(res>=80 and res<=100):
  print("Distinction")
else:
  print("Failed")
```
