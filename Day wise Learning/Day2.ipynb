```python
import pandas as pd
df=pd.read_csv("data_cleaning_dataset_1.csv")
#df.head() #Reading 1st 5 rows
print(df)
```

```python
#To know about Rows and Columns
df.shape
```

```python
#To know the duplicated values
df.duplicated()
```

```python
#To remove the duplicated values
df.drop_duplicates(inplace=True)
```

```python
#To know the null values
df.isnull()
```

```python
#To know total no of null values within each columns
df.isnull().sum()
```

```python
# Handling missing values

# Numerical Columns --> fill with midian

num_cols=["Age","Marks","Attendance_Percent"]

#with for loop
for col in num_cols:
  df[col]=df[col].fillna(df[col].median())
print(df)
```

```python
# Handling missing values

# Numerical Columns --> fill with midian

num_cols=["Age","Marks","Attendance_Percent"]

#without for loop
df[num_cols]=df[num_cols].fillna(df[num_cols].median())
print(df)
```

```python
#Categorical colums
cat_cols=["Gender","Department","City","Fee_Paid"]

# With for loop
for col in cat_cols:
  df[col]=df[col].fillna(df[col].mode()[0])
print(df)
```

```python
df.isnull().sum()
```

```python
#Categorical colums
cat_cols=["Gender","Department","City","Fee_Paid"]

# Without for loop
df[col]=df[col].fillna(df[col].mode()[0])
print(df)
```

```python
#Email --> fill missing with "unknown"

email_cols=["Email"]
df[email_cols]=df[email_cols].fillna("unknown")
print(df)

# df["Email"]=df["Email"].fillna("unknown")
# print(df)
```

```python
# Gender Standardization

df["Gender"]=df["Gender"].replace({
    "M":"Male",
    "F":"Female",
    "male":"Male",
    "female":"Female"
})
print(df)
```

```python
# Fee-Paid Standardization

df["Fee_Paid"]=df["Fee_Paid"].replace({
    "Y":"Yes",
    "N":"No"
})
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

# Label Encoding
df["Course_Level"]=le.fit_transform(df["Course_Level"])
print(df)
```

```python
# Binary encoding
df["Scholarship"]=df["Scholarship"].map({"Yes":1,"No":0})
df["Hostel"]=df["Hostel"].map({"Yes":1,"No":0})
df["Fee_Paid"]=df["Fee_Paid"].map({"Yes":1,"No":0})
print(df)
```

```python
# One-Hot encoding
df=pd.get_dummies(
    df,
    columns=["Blood_Group"],
    drop_first=True
)
print(df)
```
